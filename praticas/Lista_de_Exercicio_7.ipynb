### 3. Acurácia (Accuracy) (2/10)

A acurácia representa a proporção de amostras classificadas corretamente em relação ao total de amostras avaliadas.

Em problemas de classificação com **3 classes**, a acurácia é calculada pela soma dos elementos da diagonal principal da matriz de confusão, dividida pelo número total de amostras.

Considerando as classes 0, 1 e 2:

$$
Accuracy = \frac{VP_0 + VP_1 + VP_2}{N}
$$

Em que:

- $VP_0$: número de amostras da classe 0 classificadas corretamente.
- $VP_1$: número de amostras da classe 1 classificadas corretamente.
- $VP_2$: número de amostras da classe 2 classificadas corretamente.
- $N$: número total de amostras avaliadas.

**Atividade 2:** Para os dois bancos de dados, calcule a acurácia do classificador utilizando a função `accuracy_score()`.

Exemplo:

```

### 4. Precisão (Precision) (1/10)

A precisão indica, entre todas as amostras que o modelo classificou como pertencentes a determinada classe, quantas realmente pertencem a ela.

Em problemas de classificação com **3 classes**, calculamos a precisão individualmente para cada categoria.

Sua fórmula é:

$$
Precision_i = \frac{VP_i}{VP_i + FP_i}
$$

Em que:

- $VP_i$: número de amostras da classe $i$ classificadas corretamente.
- $FP_i$: número de amostras de outras classes classificadas incorretamente como pertencentes à classe $i$.

Por exemplo, considere a classe Peixe (0).

Se o modelo classificou 100 imagens como peixes, mas apenas 85 realmente eram peixes, sua precisão para essa classe será de 85%.

**Atividade 3:** calcule a precisão para as três classes utilizando `precision_score()`.

```

O parâmetro `average=None` retorna uma métrica individual para cada classe, na ordem dos rótulos.

### 5. Revocação ou Sensibilidade (Recall) (1/10)

A revocação indica a capacidade do modelo de identificar corretamente as amostras que realmente pertencem a determinada classe.

Em problemas de classificação com **3 classes**, calculamos a revocação individualmente para cada categoria.

Sua fórmula é:

$$
Recall_i = \frac{VP_i}{VP_i + FN_i}
$$

Em que:

- $VP_i$: número de amostras da classe $i$ classificadas corretamente.
- $FN_i$: número de amostras da classe $i$ classificadas incorretamente como pertencentes a outras classes.

Por exemplo, considere novamente a classe Peixe (0).

Se existem 100 imagens de peixes no conjunto de teste e o modelo identificou corretamente 85 delas, sua revocação será de 85%.

Diferentemente da precisão, que considera as previsões positivas realizadas pelo modelo, a revocação considera todas as amostras que realmente pertencem à classe.

O parâmetro `average=None` retorna uma revocação individual para cada classe, na ordem dos rótulos.

### 6. F1-score (1/10)

O F1-score é uma métrica que combina precisão e revocação por meio da média harmônica entre elas.

Em problemas de classificação com **3 classes**, calculamos o F1-score individualmente para cada categoria.

Sua fórmula é:

$$
F1_i = 2 \cdot \frac{Precision_i \cdot Recall_i}{Precision_i + Recall_i}
$$

Em que:

- $Precision_i$: precisão da classe $i$.
- $Recall_i$: revocação da classe $i$.

Essa métrica é especialmente útil quando desejamos avaliar conjuntamente a capacidade do modelo de evitar falsos positivos e falsos negativos.

Um F1-score elevado indica que o modelo apresenta bons resultados tanto em precisão quanto em revocação.

### 7. Acurácia Balanceada (Balanced Accuracy) (1/10)

A acurácia balanceada é calculada pela média aritmética da revocação obtida em cada classe.

Considerando um problema de classificação com **3 classes**, sua fórmula é:

$$
Balanced\ Accuracy =
\frac{Recall_0 + Recall_1 + Recall_2}{3}
$$

Em que:

- $Recall_0$: revocação da classe 0 (Peixe).
- $Recall_1$: revocação da classe 1 (Gato).
- $Recall_2$: revocação da classe 2 (Cachorro).

Essa métrica atribui a mesma importância a todas as classes, independentemente de sua frequência no conjunto de dados.

Em conjuntos balanceados, a acurácia balanceada coincide com a acurácia convencional.

**Atividade 7:** calcule a acurácia balanceada.

```python

    df["y_true"],
    df["y_pred"],
)

print("Balanced Accuracy:", balanced_acc)
```
